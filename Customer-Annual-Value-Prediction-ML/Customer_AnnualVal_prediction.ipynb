```python
import pandas as pd

from sklearn.model_selection import train_test_split

from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder

from sklearn.linear_model import LinearRegression

from sklearn.tree import DecisionTreeRegressor

from sklearn.ensemble import RandomForestRegressor

from sklearn.svm import SVR

from sklearn.metrics import mean_squared_error
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import r2_score

import matplotlib.pyplot as plt
import seaborn as sns

import pickle
```

```python
df=pd.read_csv("Customer_Regression_Practice_20000.csv")
df.head()
```

```python
df.info()
```

```python
l1=LabelEncoder()
l2=LabelEncoder()
df['Gender']=l1.fit_transform(df['Gender'])
df['Plan_Type']=l2.fit_transform(df['Plan_Type'])
```

```python
ind=df.iloc[:,0:14]
dep=df.iloc[:,-1]
```

```python
ind.columns
```

```python
ind_train,ind_test,dep_train,dep_test=train_test_split(ind,dep,test_size=0.2,random_state=42)

display(ind_train.shape,dep_train.shape,dep_test.shape,dep_test.shape)
```

```python
#LINEARREGRESSION
m1=LinearRegression()
m1.fit(ind_train,dep_train)
```

```python
pred=m1.predict(ind_test)
pred
```

```python
#DecisionTreeRegressor()
from sklearn.tree import DecisionTreeRegressor

m2 = DecisionTreeRegressor(criterion='squared_error')

m2.fit(ind_train, dep_train)

```

```python
pred = m2.predict(ind_test)

print(pred)
```

```python
#RandomForestRegressor
m3=RandomForestRegressor()
m3.fit(ind_train,dep_train)
```

```python

pred = m3.predict(ind_test)

print(pred)
```

```python
m4=SVR()
m4.fit(ind_train,dep_train)
```

```python
pred=m4.predict(ind_test)
print(pred)
```

```python
age = int(input("Enter the age: "))

gender = input("Enter the gender: ")

plan_type = input("Enter the plan type: ")

tenure_months = int(input("Enter the tenure months: "))

monthly_usage_hours = float(input("Enter monthly usage hours: "))

support_tickets = int(input("Enter support tickets: "))

satisfaction_score = float(input("Enter satisfaction score: "))

monthly_charges = float(input("Enter monthly charges: "))

data_usage_gb = float(input("Enter data usage in GB: "))

payment_delay_days = int(input("Enter payment delay days: "))

contract_months = int(input("Enter contract months: "))

avg_login_days = float(input("Enter average login days: "))

service_count = int(input("Enter service count: "))

discount_percent = float(input("Enter discount percentage: "))


# Label Encoding
Gender = l1.transform([gender])[0]

Plan_Type = l2.transform([plan_type])[0]


# Create input DataFrame
input_data = pd.DataFrame([[
    age,
    Gender,
    Plan_Type,
    tenure_months,
    monthly_usage_hours,
    support_tickets,
    satisfaction_score,
    monthly_charges,
    data_usage_gb,
    payment_delay_days,
    contract_months,
    avg_login_days,
    service_count,
    discount_percent
]], columns=[
    'Age',
    'Gender',
    'Plan_Type',
    'Tenure_Months',
    'Monthly_Usage_Hours',
    'Support_Tickets',
    'Satisfaction_Score',
    'Monthly_Charges',
    'Data_Usage_GB',
    'Payment_Delay_Days',
    'Contract_Months',
    'Avg_Login_Days',
    'Service_Count',
    'Discount_Percent'
])


# Predictions
result1 = m1.predict(input_data)
result2 = m2.predict(input_data)
result3 = m3.predict(input_data)
result4 = m4.predict(input_data)

print("Model 1 Prediction:", result1[0])
print("Model 2 Prediction:", result2[0])
print("Model 3 Prediction:", result3[0])
print("Model 4 Prediction:", result4[0])
```

```python
pickle.dump(m1,open("model1.pkl","wb"))
pickle.dump(m2,open("model2.pkl","wb"))
pickle.dump(m3,open("model3.pkl","wb"))
pickle.dump(m4,open("model4.pkl","wb"))

pickle.dump(l1, open("gender_encoder.pkl", "wb"))
pickle.dump(l2, open("plan_encoder.pkl", "wb"))
print("All models saved successfully!")
```
